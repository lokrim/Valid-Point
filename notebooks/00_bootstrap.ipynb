```python
from valid_point.provenance import (notebook_context, bootstrap_tables, export_table, structure_figure, export_roundtrip, output_stem, format_source_tree, source_inventory)
from pathlib import Path
import json

root, run_dir, config, provenance = notebook_context()
print("Repository:", root)
print("Run:", run_dir.name)
print("Configuration (exact):")
print(json.dumps(config, indent=2, sort_keys=True))
print("Output naming example:", output_stem("T00_environment", provenance))
```

## T00_environment — interpreter and package origins

Observed package versions are structural environment evidence. They are not research measurements.

```python
environment_rows, scaffold_rows, source_hash = bootstrap_tables(root, config)
for row in environment_rows:
    print(f"{row['item']}: {row['value']} [{row['state']}]")
print("Source snapshot SHA-256:", source_hash)
environment_paths = export_table(environment_rows, "T00_environment", root / "reports/tables", provenance)
print("T00_environment exports:", *environment_paths, sep="\n")
```

## T00_scaffold and F00_structure — repository inventory and flow

The full small source inventory is shown with SHA-256 hashes. The diagram depicts source → package → fresh kernel → evidence exports; it does not depict a research algorithm.

```python
print(format_source_tree(source_inventory(root)[0]))
for row in scaffold_rows:
    print(f"{row['path']}  {row['sha256']}  {row['state']}")
scaffold_paths = export_table(scaffold_rows, "T00_scaffold", root / "reports/tables", provenance)
figure_paths = structure_figure(root / "reports/figures", provenance)
print("T00_scaffold exports:", *scaffold_paths, sep="\n")
print("F00_structure exports:", *figure_paths, sep="\n")
from IPython.display import Image, display
display(Image(filename=str(figure_paths[2])))
```

## Evidence-based gate

The test report was produced by the runner before this notebook. The final gate file and manifest are written after this notebook completes, so an execution failure cannot claim a pass.

```python
test_report = json.loads((run_dir / "test_results.json").read_text(encoding="utf-8"))
print("Test command:", " ".join(test_report["command"]))
print(test_report["output"])
assert test_report["exit_code"] == 0, "bootstrap tests failed"
assert roundtrip["roundtrip"] == "pass" and roundtrip["missing_provenance"].startswith("rejected:")
assert len(environment_paths) == 3 and len(scaffold_paths) == 3 and len(figure_paths) == 4
print("Notebook structural gate: PASS. Research science: not run. G1 scientific contracts: pending S01.")
```
