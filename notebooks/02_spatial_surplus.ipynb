# S02 — Counts in externally fixed regions

Track: GT-free raw hand fixtures. Split: development only. Fitting, scoring and calibration await a separately invoked S03.

Three receiver-owned 5 m × 5 m tiles, height slab [-1, 1) m, are frozen at time zero before inspecting any sender. Bounds are lower-inclusive and upper-exclusive; adjacent boundaries belong to one tile. All region IDs remain present, including zero-count C. Counts outside the region union are retained. This small partial region universe tests raw geometry, not the later S04 full-ROI or oracle-gap study.

The baseline counts [2,1,0] become [3,2,0] after literal point addition. Dense legitimate returns produce [9,0,0]; changed content produces [0,3,0]. Rearrangement leaves counts unchanged and removal lowers them. None of these observations defines a trust reward or penalty. Visibility stays unknown, including for parsed empty clouds; absence yields null counts.

Pass criterion: exact counts, boundary/transform/unknown tests and literal-addition monotonicity pass. No symmetric closeness reward, point-share factor, expected-count reference or normalized surplus is computed.
