# S02 — Availability and independent eligibility

Track: GT-free raw hand fixtures. Split: development only. Fitting, scoring and calibration await a separately invoked S03.

The decision unit is sender/frame/deadline. Freshness is deadline minus source timestamp in seconds, with a declared 200 ms maximum. Availability, parsed payload, transform validity, independent health and membership are separate fields. A stale source, missing cloud, sensor fault or malformed packet does not establish intent.

Known raw counts may coexist with unknown health or membership. Overall eligibility is conservative and remains unknown/invalid until every declared prerequisite is known. Late packet source time and receipt are shown only as synthetic timeline diagnostics; their content cannot enter measurements at the deadline. Health is sourced from a hand-defined independent receiver observation, never inferred from sender self-report. Future context is unavailable.

Pass criterion: the explicit independent statuses match every hand case. Regions are **not applicable** to this availability factor; it consumes only receiver context and message/parse metadata.

```python
import json, os, sys
from pathlib import Path
from IPython.display import display, Markdown, Image, JSON
from valid_point.evidence import load_raw_config, export_demo
from valid_point.provenance import sha256_file
root = Path(os.environ['VP_REPO_ROOT'])
run_dir = Path(os.environ['VP_RUN_DIR'])
config = load_raw_config(run_dir / 'config.json')
provenance = dict(run_id=run_dir.name, stage='S02', track=config['track'],
                  segment=config['segment'], config_sha256=sha256_file(run_dir/'config.json'),
                  source_sha256=os.environ['VP_SOURCE_SHA256'], notebook_section='02_availability')
print('Fresh kernel:', sys.executable)
print(json.dumps(provenance, indent=2))
display(JSON({k:v for k,v in config.items() if not k.endswith('_cases')}, expanded=True))
print('Reference/calibration: not fitted. Normalized evidence, score, alarm, allocation: not run.')
```
